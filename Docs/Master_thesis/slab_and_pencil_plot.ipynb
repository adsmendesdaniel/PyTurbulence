Import/install libraries:

```python
import numpy as np
import matplotlib.pyplot as plt
```

Slab plot:

```python
# domain size
Nx, Ny, Nz = 8, 8, 8

# voxel grid
x, y, z = np.indices((Nx, Ny, Nz))
voxelarray = np.ones((Nx, Ny, Nz), dtype=bool)

# number of slabs
n_slabs = 4
slab_size = Nx // n_slabs

# color array
colors = np.empty(voxelarray.shape, dtype=object)

# color map for slabs
slab_colors = [
    '#1f77b4',  # blue
    '#ff7f0e',  # orange
    '#2ca02c',  # green
    '#d62728',  # red
]

for i in range(n_slabs):
    x_start = i * slab_size
    x_end = (i + 1) * slab_size
    colors[x_start:x_end, :, :] = slab_colors[i]

# plot
fig = plt.figure(figsize=(6.4, 6.4))
ax = fig.add_subplot(projection='3d')

ax.voxels(
    voxelarray,
    facecolors=colors,
    edgecolor='k'
)

# axis scaling
ticks = np.linspace(0, Nx, 3)
tick_labels = [r'$0$', r'$\pi$', r'$2\pi$']

ax.set_xticks(ticks)
ax.set_yticks(ticks)
ax.set_zticks(ticks)

ax.set_xticklabels(tick_labels)
ax.set_yticklabels(tick_labels)
ax.set_zticklabels(tick_labels)

# bold axis labels (x and y)
ax.set_xlabel(r'$\mathbf{x}$', labelpad=0)
ax.set_ylabel(r'$\mathbf{y}$', labelpad=0)
ax.set_zlabel(r'$\mathbf{z}$', labelpad=-1)

# save figure
plt.savefig(
    "slab_decomposition_mesh.png",
    dpi=300,
    #bbox_inches='tight',
    pad_inches=0.05
)

plt.show()
```

Pencil plot:

```python
# domain size
Nx, Ny, Nz = 8, 8, 8

# voxel grid
x, y, z = np.indices((Nx, Ny, Nz))
voxelarray = np.ones((Nx, Ny, Nz), dtype=bool)

# pencil decomposition: n_px × n_py pencils
n_px, n_py = 2, 2
px_size = Nx // n_px
py_size = Ny // n_py

# color array
colors = np.empty(voxelarray.shape, dtype=object)

# color map for pencils
pencil_colors = [
    '#1f77b4',  # blue
    '#ff7f0e',  # orange
    '#2ca02c',  # green
    '#d62728',  # red
]

# assign colors to pencils
c = 0
for i in range(n_px):
    for j in range(n_py):
        x_start = i * px_size
        x_end   = (i + 1) * px_size
        y_start = j * py_size
        y_end   = (j + 1) * py_size

        colors[x_start:x_end, y_start:y_end, :] = pencil_colors[c]
        c += 1

# plot
fig = plt.figure(figsize=(6.4, 6.4))
ax = fig.add_subplot(projection='3d')

ax.voxels(
    voxelarray,
    facecolors=colors,
    edgecolor='k'
)

# axis scaling
ticks = np.linspace(0, Nx, 3)
tick_labels = [r'$0$', r'$\pi$', r'$2\pi$']

ax.set_xticks(ticks)
ax.set_yticks(ticks)
ax.set_zticks(ticks)

ax.set_xticklabels(tick_labels)
ax.set_yticklabels(tick_labels)
ax.set_zticklabels(tick_labels)

# bold axis labels
ax.set_xlabel(r'$\mathbf{x}$', labelpad=0)
ax.set_ylabel(r'$\mathbf{y}$', labelpad=0)
ax.set_zlabel(r'$\mathbf{z}$', labelpad=-1)

# save figure
plt.savefig(
    "pencil_decomposition_mesh.png",
    dpi=300,
    pad_inches=0.05
)

plt.show()
```
