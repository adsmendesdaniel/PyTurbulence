# 2D early tests:

## ---2**8--- Test for the perlin noise until T=100

```python
# Hardware color mapping
hardware_colors = {
    "Local Machine": "C0",
    "Google Colab": "C1",         
    #"SD Login Node": "C2",
    }

# Data assignment
data = [
    ("1", (31+30+30)/3, "Local Machine"),
    ("2", (31+30+30)/3, "Local Machine"),
    ("4", (14+13+13)/3, "Local Machine"),
    ("8", (27+25+26)/3, "Local Machine"),

    ("1 (CPU)", (32+32+33)/3, "Google Colab"),
    ("1 (Cupy)", (23+22+23)/3, "Google Colab"),
    ("1 (CupyX/SciPy)", (22.46+22.53+22.53)/3, "Google Colab"),
]

# Unpack data
labels  = [d[0] for d in data]
values  = [d[1] for d in data]
classes = [d[2] for d in data]
colors  = [hardware_colors[c] for c in classes]

# Layout logic
BAR_HEIGHT = 0.6
FIG_WIDTH = 10
fig_height = BAR_HEIGHT * len(data) + 1.2

fig, ax = plt.subplots(figsize=(FIG_WIDTH, fig_height), dpi=100)

# Horizontal bars
y_pos = range(len(data))

bars = ax.barh(
    y_pos,
    values,
    height=BAR_HEIGHT,
    color=colors,
    edgecolor="black"
)

# Value labels on bars
for bar, value in zip(bars, values):
    ax.text(
        value + max(values) * 0.01,
        bar.get_y() + bar.get_height() / 2,
        f"{value:.2f}",
        va="center",
        fontsize=10
    )

# Axis labels & ticks
ax.set_xlabel("Execution time [s]")
ax.set_ylabel("Number of processes")

ax.set_yticks(y_pos)
ax.set_yticklabels(labels)

# Reverse vertical order (top-to-bottom)
ax.invert_yaxis()

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Legend
legend_elements = [
    Patch(facecolor=hardware_colors[name],
          edgecolor="black",
          label=name)
    for name in hardware_colors
]

ax.legend(
    handles=legend_elements,
)

# Grid & layout
plt.grid(alpha=0.4, axis="x")
plt.tight_layout()
plt.savefig("performance_barplot_256_2d_perlin_noise_.png", dpi=300)
plt.show()
```
