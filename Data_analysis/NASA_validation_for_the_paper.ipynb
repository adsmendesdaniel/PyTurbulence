# Import and/or install libraries: 

```python
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt
```

# Read data

```python
# NASA data:
NASA_data_512 = np.loadtxt(
    "../Data/NASA_workshop_validation/C3.3_datafiles/spectral_Re1600_512.gdiag"
)
time_NASA_512 = NASA_data_512[:, 0]
energy_NASA_512 = NASA_data_512[:, 1]
dissipation_rate_NASA_512 = NASA_data_512[:, 2]
enstrophy_NASA_512 = NASA_data_512[:, 3]

# 64^3 data:
energy_data_64 = np.loadtxt("../Data/metrics_64/energy_timeseries.txt")
enstrophy_data_64 = np.loadtxt("../Data/metrics_64/enstrophy_timeseries.txt")
dissipation_rate_64 = np.column_stack(
    (energy_data_64[:, 0], -np.gradient(energy_data_64[:, 1], energy_data_64[:, 0]))
)

# 64^3 data with 1024^3 timestep:
energy_data_64_timestep_1024 = np.loadtxt(
    "../Data/metrics_64_timestep_1024/energy_timeseries.txt"
)
enstrophy_data_64_timestep_1024 = np.loadtxt(
    "../Data/metrics_64_timestep_1024/enstrophy_timeseries.txt"
)
dissipation_rate_64_timestep_1024 = np.column_stack(
    (
        energy_data_64_timestep_1024[:, 0],
        -np.gradient(
            energy_data_64_timestep_1024[:, 1], energy_data_64_timestep_1024[:, 0]
        ),
    )
)

# 128^3 data:
energy_data_128 = np.loadtxt("../Data/metrics_128/energy_timeseries.txt")
enstrophy_data_128 = np.loadtxt("../Data/metrics_128/enstrophy_timeseries.txt")
dissipation_rate_128 = np.column_stack(
    (energy_data_128[:, 0], -np.gradient(energy_data_128[:, 1], energy_data_128[:, 0]))
)

# 128^3 data with 1024^3 timestep:
energy_data_128_timestep_1024 = np.loadtxt(
    "../Data/metrics_128_timestep_1024/energy_timeseries.txt"
)
enstrophy_data_128_timestep_1024 = np.loadtxt(
    "../Data/metrics_128_timestep_1024/enstrophy_timeseries.txt"
)
dissipation_rate_128_timestep_1024 = np.column_stack(
    (
        energy_data_128_timestep_1024[:, 0],
        -np.gradient(
            energy_data_128_timestep_1024[:, 1], energy_data_128_timestep_1024[:, 0]
        ),
    )
)

# 256^3 data:
energy_data_256 = np.loadtxt("../Data/metrics_256/energy_timeseries.txt")
enstrophy_data_256 = np.loadtxt("../Data/metrics_256/enstrophy_timeseries.txt")
dissipation_rate_256 = np.column_stack(
    (energy_data_256[:, 0], -np.gradient(energy_data_256[:, 1], energy_data_256[:, 0]))
)

# 512³ data:
energy_data_512 = np.loadtxt("../Data/metrics_512/energy_timeseries.txt")
enstrophy_data_512 = np.loadtxt("../Data/metrics_512/enstrophy_timeseries.txt")
dissipation_rate_512 = np.column_stack(
    (energy_data_512[:, 0], -np.gradient(energy_data_512[:, 1], energy_data_512[:, 0]))
)
```

# Energy:

```python
sns.set_theme(style="ticks", context="notebook", palette="colorblind")

plt.figure(figsize=(6.4 * 0.8, 4.8 * 0.8), dpi=300)

plt.plot(
    energy_data_64[:, 0],
    energy_data_64[:, 1],
    label=rf"$64^3$",
    color="C0",
    linewidth=2.0,
)
plt.plot(
    energy_data_128[:, 0],
    energy_data_128[:, 1],
    label=rf"$128^3$",
    color="C1",
    linewidth=2.0,
)
plt.plot(
    energy_data_256[:, 0],
    energy_data_256[:, 1],
    label=rf"$256^3$",
    color="C2",
    linewidth=2.0,
)
plt.plot(
    energy_data_512[:, 0],
    energy_data_512[:, 1],
    label=rf"$512^3$",
    color="tab:red",
    linewidth=2.0,
)
plt.plot(
    time_NASA_512,
    energy_NASA_512,
    label="NASA Data",
    color="k",
    linewidth=2.0,
    linestyle="--",
    alpha=0.8,
)

plt.xlabel(rf"$t$")
plt.ylabel("TKE")

plt.grid(alpha=0.5)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(6.4, 4.8), dpi=150)

plt.plot(
    energy_data_64[:, 0],
    energy_data_64[:, 1],
    label="Energy (64x64)",
    color="C0",
    linewidth=2.5,
)
plt.plot(
    energy_data_128[:, 0],
    energy_data_128[:, 1],
    label="Energy (128x128)",
    color="C1",
    linewidth=2.5,
)
plt.plot(
    energy_data_256[:, 0],
    energy_data_256[:, 1],
    label="Energy (256x256)",
    color="C2",
    linewidth=2.5,
)
plt.plot(
    time_NASA_512,
    energy_NASA_512,
    label="Energy (512x512)",
    color="k",
    linewidth=2.5,
    linestyle="--",
    alpha=0.6,
)

plt.xlabel("Timestep")
plt.ylabel("Enstrophy")

plt.xlim(0, 10)
plt.legend()
plt.show()
```

# Energy dissipation:

```python
sns.set_theme(style="ticks", context="paper", palette="colorblind")

plt.figure(figsize=(6.4 * 0.8, 4.8 * 0.8), dpi=300)

plt.plot(
    dissipation_rate_64[:, 0],
    dissipation_rate_64[:, 1],
    label=rf"$64^3$",
    color="C0",
    linewidth=2.0,
)
plt.plot(
    dissipation_rate_128[:, 0],
    dissipation_rate_128[:, 1],
    label=rf"$128^3$",
    color="C1",
    linewidth=2.0,
)
plt.plot(
    dissipation_rate_256[:, 0],
    dissipation_rate_256[:, 1],
    label=rf"$256^3$",
    color="C2",
    linewidth=2.0,
)
plt.plot(
    dissipation_rate_512[:, 0],
    dissipation_rate_512[:, 1],
    label=rf"$512^3$",
    color="tab:red",
    linewidth=2.0,
)
plt.plot(
    time_NASA_512,
    dissipation_rate_NASA_512,
    label="NASA Data",
    color="k",
    linewidth=2.0,
    linestyle="--",
    alpha=0.8,
)

plt.xlabel(rf"$t$")
plt.ylabel(rf"$\epsilon$")

plt.legend()
plt.grid(alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(6.4, 4.8), dpi=150)

plt.plot(dissipation_rate_64[:, 0], dissipation_rate_64[:, 1], label='Dissipation Rate (64x64)', color='C0', linewidth=2.5)
plt.plot(dissipation_rate_128[:, 0], dissipation_rate_128[:, 1], label='Dissipation Rate (128x128)', color='C1', linewidth=2.5)
plt.plot(dissipation_rate_256[:, 0], dissipation_rate_256[:, 1], label='Dissipation Rate (256x256)', color='C2', linewidth=2.5)

plt.plot(time_NASA_512, dissipation_rate_NASA_512, label='Dissipation Rate (512x512)', color='k', linewidth=2.5, linestyle='--', alpha=0.6)

plt.xlabel('Timestep')
plt.ylabel('Dissipation Rate')

plt.xlim(0, 10) 
plt.legend()
plt.show()
```
