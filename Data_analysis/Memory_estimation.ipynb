# Estimation with the ENMC paper formula:

```python
grid_size_sd_login = [2**1, 2**2, 2**3, 2**4, 2**5, 2**6, 2**7, 2**8, 2**9]
allocated_memory_sd_login = np.array([431, 437, 437, 437, 439, 461, 629, 1777, 10973])*4

grid_size_kaggle = [2**1, 2**2, 2**3, 2**4, 2**5, 2**6, 2**7, 2**8]
allocated_memory_kaggle = np.array([188, 188, 188, 188, 192, 248, 564, 2972])*4

```

```python
# Given data
grid_size_sd_login = [2**1, 2**2, 2**3, 2**4, 2**5, 2**6, 2**7, 2**8, 2**9]
allocated_memory_sd_login = np.array([431, 437, 437, 437, 439, 461, 629, 1777, 10973])*4

grid_size_kaggle = [2**1, 2**2, 2**3, 2**4, 2**5, 2**6, 2**7, 2**8]
allocated_memory_kaggle = np.array([188, 188, 188, 188, 192, 248, 564, 2972])*4

# Model function: M_total = M0 + alpha * N^3
def memory_model(N, M0, alpha):
    return M0 + alpha * N**3

# Fit for sd_login data
params_sd, cov_sd = curve_fit(memory_model, grid_size_sd_login, allocated_memory_sd_login)
M0_sd, alpha_sd = params_sd

# Predict memory for N=2^10
N_predict = 2**10
memory_predict_sd = memory_model(N_predict, M0_sd, alpha_sd)

print(f"SD_Login fit parameters: M0 = {M0_sd:.2f}, alpha = {alpha_sd:.6f}")
print(f"Predicted memory for N=2^10 (SD_Login): {memory_predict_sd:.2f}")

# Fit for kaggle data
params_kaggle, cov_kaggle = curve_fit(memory_model, grid_size_kaggle, allocated_memory_kaggle)
M0_kaggle, alpha_kaggle = params_kaggle

memory_predict_kaggle = memory_model(N_predict, M0_kaggle, alpha_kaggle)

print(f"Kaggle fit parameters: M0 = {M0_kaggle:.2f}, alpha = {alpha_kaggle:.6f}")
print(f"Predicted memory for N=2^10 (Kaggle): {memory_predict_kaggle:.2f}")



# PLOT:
sns.set_theme(style="ticks", context='paper', palette='colorblind')

# Plot of log space:
plt.figure(figsize=(6.4*0.8, 4.8*0.8), dpi=150)

plt.scatter(grid_size_sd_login, allocated_memory_sd_login, label='SD_Login Data', color='blue')

N_fit = np.linspace(2, 2**10, 100)
plt.plot(N_fit, memory_model(N_fit, *params_sd), label='SD_Login Fit', color='blue', linestyle='--')

plt.xscale('log', base=2)
plt.yscale('log')
plt.xlabel('Grid Size N (log2 scale)')
plt.ylabel('Allocated Memory (units)')
plt.title('Memory Allocation Fit')
plt.legend()
plt.grid(True, which="both", ls="--")
plt.show()
```

## Plot for the paper:

```python
# Given data
grid_size_sd_login = [2**1, 2**2, 2**3, 2**4, 2**5, 2**6, 2**7, 2**8, 2**9]
allocated_memory_sd_login = (
    np.array([431, 437, 437, 437, 439, 461, 629, 1777, 10973]) * 4
)


# Model function: M_total = M0 + alpha * N^3
def memory_model(N, M0, alpha):
    return M0 + alpha * N**3


# Fit for sd_login data
params_sd, cov_sd = curve_fit(
    memory_model, grid_size_sd_login, allocated_memory_sd_login
)
M0_sd, alpha_sd = params_sd

# Predict memory for N=2^10
N_predict = 2**10
memory_predict_sd = memory_model(N_predict, M0_sd, alpha_sd)

print(f"SD_Login fit parameters: M0 = {M0_sd:.2f}, alpha = {alpha_sd:.6f}")
print(f"Predicted memory for N=2^10 (SD_Login): {memory_predict_sd:.2f}")


# PLOT:
sns.set_theme(style="ticks", context="notebook", palette="colorblind")

# Plot of log space:
plt.figure(figsize=(6.4 * 0.8, 4.8 * 0.8), dpi=300)

# Add predicted values to the arrays so we can plot them
allocated_memory_sd_login = np.concatenate(
    [allocated_memory_sd_login, [memory_predict_sd]]
)
grid_size_sd_login = np.concatenate([grid_size_sd_login, [2**10]])

# Kaggle 1x P100
plt.axhline(y=16, color="grey", linestyle="--", alpha=0.7)
plt.text(
    2**1, 16, "Kaggle (1x P100)", ha="left", va="bottom", color="grey", fontsize=10
)

# Kaggle 2x T4
plt.axhline(y=30, color="grey", linestyle="--", alpha=0.7)
plt.text(2**1, 30, "Kaggle (2x T4)", ha="left", va="bottom", color="grey", fontsize=10)

# SD Login node + GDL
plt.axhline(y=32 * 4, color="grey", linestyle="--", alpha=0.7)
plt.text(
    2**1,
    32 * 4 - 35,
    "SD Login Node (4x V100-PCIe)",
    ha="left",
    va="bottom",
    color="grey",
    fontsize=10,
)
plt.text(
    2**1,
    16 * 8,
    "SD GDL Node (8x V100-SXM2)",
    ha="left",
    va="bottom",
    color="grey",
    fontsize=10,
)

plt.scatter(
    grid_size_sd_login,
    allocated_memory_sd_login / 1024,
    label="Measurements",
    color="C4",
)

N_fit = np.linspace(2, 2**10, 100)
plt.plot(
    N_fit,
    memory_model(N_fit, *params_sd) / 1024,
    label=rf"Fit: M = $M_0$ + $\alpha \cdot N^3$",
    linestyle="--",
    color="C4",
    markersize=5,
    zorder=-1
)

plt.scatter(
    grid_size_sd_login[-1],
    allocated_memory_sd_login[-1] / 1024,
    label="Prediction",
    color="k",
)

# Axes config
ax = plt.gca()

# keep your x as base-2 log with 2^k labels
ax.set_xscale("log", base=2)
ax.set_xticks(grid_size_sd_login)
ax.set_xticklabels(
    [f"$2^{{{int(np.log2(n))}}}$" for n in grid_size_sd_login], rotation=45
)

# y as base-10 log with default ticks
ax.set_yscale("log", base=2)

# no grid lines at all
ax.grid(False)

plt.xlabel("Grid Resolution")
plt.ylabel("Required GPU memory (GiB)")
plt.legend(loc=(0.02, 0.125))
plt.grid(alpha=0.5)
plt.tight_layout()
plt.show()
```
