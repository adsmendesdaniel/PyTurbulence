# Errors:


```python
import scipy.interpolate as interpolate


def interpolate_to_common_grid(time1, data1, time2, data2, method="linear"):
    """
    Interpolate two datasets to a common time grid for comparison
    """
    # Find common time range
    t_start = max(time1.min(), time2.min())
    t_end = min(time1.max(), time2.max())

    # Create common time grid (use the finer resolution)
    dt1 = np.mean(np.diff(time1))
    dt2 = np.mean(np.diff(time2))
    dt_common = min(dt1, dt2)

    t_common = np.arange(t_start, t_end + dt_common, dt_common)

    # Interpolate both datasets
    f1 = interpolate.interp1d(
        time1, data1, kind=method, bounds_error=False, fill_value="extrapolate"
    )
    f2 = interpolate.interp1d(
        time2, data2, kind=method, bounds_error=False, fill_value="extrapolate"
    )

    data1_interp = f1(t_common)
    data2_interp = f2(t_common)

    return t_common, data1_interp, data2_interp


def calculate_errors(y_true, y_pred):
    """
    Calculate various error metrics between two datasets
    """
    rmse_val = np.sqrt(np.mean((y_true - y_pred) ** 2))
    mae = np.mean(np.abs(y_true - y_pred))
    relative_error = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    l2_norm_error = np.linalg.norm(y_true - y_pred) / np.linalg.norm(y_true)
    max_error = np.max(np.abs(y_true - y_pred))

    return {
        "RMSE": rmse_val,
        "MAE": mae,
        "Relative_Error_%": relative_error,
        "L2_Norm_Error": l2_norm_error,
        "Max_Error": max_error,
    }
```

**Energy**

```python
# Calculate_errors

resolutions = {
    "64x64": energy_data_64,
    "128x128": energy_data_128,
    "256x256": energy_data_256,
    "512x512": energy_data_512,
}

for label, data in resolutions.items():
    time_sim = data[:, 0]
    enstrophy_sim = data[:, 1]

    t_common, y_sim, y_nasa = interpolate_to_common_grid(
        time_sim, enstrophy_sim, time_NASA_512, energy_NASA_512
    )
    errors = calculate_errors(y_nasa, y_sim)

    print(f"\n--- Errors for resolution {label} ---")
    for metric, value in errors.items():
        print(f"{metric}: {value:.6f}")

    fig, ax1 = plt.subplots(figsize=(6.4, 4.8), dpi=100)

    # Plot simulation and NASA reference curves on left y-axis
    ax1.plot(t_common, y_sim, label=f"Simulation ({label})", color='C0')
    ax1.plot(t_common, y_nasa, label=f"NASA Reference ({label})", linestyle="--", color='C1')
    ax1.set_xlabel("Time")
    ax1.set_ylabel("Energy", color='black')
    ax1.tick_params(axis='y', labelcolor='black')

    # Calculate absolute difference
    difference = np.abs(y_sim - y_nasa)

    # Create second y-axis sharing the same x-axis
    ax2 = ax1.twinx()

    # Shade absolute difference on right y-axis
    ax2.fill_between(
        t_common,
        0,
        difference,
        color="gray",
        alpha=0.3,
        label="Absolute Difference",
    )
    ax2.set_ylabel("Absolute Difference", color='gray')
    ax2.tick_params(axis='y', labelcolor='gray')

    # Combine legends from both axes
    lines_1, labels_1 = ax1.get_legend_handles_labels()
    lines_2, labels_2 = ax2.get_legend_handles_labels()
    ax1.legend(lines_1 + lines_2, labels_1 + labels_2)

    plt.show()
```
