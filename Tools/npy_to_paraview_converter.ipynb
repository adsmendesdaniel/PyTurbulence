# Import/Install libraries:

```python
import numpy as np
import pyvista as pv
```

# Read data:

```python
data = np.load("../Data/metrics_512/vorticity_rank_1_t8.npy")

print(data.shape)
```

# Convert and save:

# VTI

```python
spatial_dims = data.shape[1:]  # (256, 256, 256)

grid = pv.ImageData()
grid.dimensions = np.array(spatial_dims) + 1  # Must be +1 in each dim
grid.spacing = (1, 1, 1)  # adjust if your grid spacing is different
grid.origin = (0, 0, 0)

# PyVista expects vector data as (n_points, 3) or (n_cells, 3)
vector = np.transpose(data, (1, 2, 3, 0))  # Reshape to (256, 256, 256, 3)
grid.cell_data["velocity"] = vector.reshape(-1, 3, order="F")

# Save to .vti format
grid.save("vorticity_data_t8_512_rank_1.vti")
```
